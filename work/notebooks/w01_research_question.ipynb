## 1. My lane (or freestyle) and why

**Provisional Lane:** Refresh / Content Opportunity Scoring (Lane 2)

I chose this lane because content decay is a primary challenge in SEO. The starter dataset provides rich signals (impressions, clicks, age, and update history) that allow for a more nuanced prioritization of content refreshes than simple time-based rules. By identifying pages that are losing momentum but still have high search exposure, we can help editors focus their limited capacity on the highest-impact updates.

## 2. The question: decision, action, cost of a wrong call

**Research Question:** Can we identify which high-exposure content pages are entering a period of decline early enough to prioritize them for a refresh before they lose significant organic traffic?

**Unit of Analysis:** A single pseudonymized content item (page).

**The Decision:** Which specific pages should an SEO or content editor review and update this week?

**The Action:** The editor reviews the prioritized page and performs a content refresh (e.g., updating facts, expanding sections, or improving intent match).

**The Output:** A ranked queue of content items with reason codes and confidence scores.

**Cost of a Wrong Call:**
*   **False Positive (Waste):** An editor spends several hours refreshing a page that was actually stable or not in need of update, wasting creative resources that could have been used elsewhere.
*   **False Negative (Miss):** A high-traffic page continues to decay unnoticed for months, leading to a sustained loss of search visibility and potential revenue/leads.

**Why Data/ML Helps:** Standard SEO rules (e.g., "refresh every 6 months") are too broad. They ignore whether a page is actually losing traffic or if it has any search demand. ML can combine multiple signals—like declining CTR alongside steady impressions—to find subtle patterns of decay that a human or a simple rule would miss across thousands of pages.

## 3. Quick look at the data (2-3 real numbers)

The following code loads the starter dataset and calculates three key statistics that justify this research direction.

```python
import pandas as pd
import os

# Adjust path for repository structure
data_path = "../../data/raw/content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    data_path = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

total_records = len(df)
declining_count = (df['trend_direction'] == 'down').sum()
declining_pct = (declining_count / total_records) * 100
avg_imp_declining = df[df['trend_direction'] == 'down']['impressions_90d'].mean()

print(f"Fact 1: Total pages analyzed: {total_records:,}")
print(f"Fact 2: Percentage of pages currently in a 'down' trend: {declining_pct:.1f}%")
print(f"Fact 3: Average 90-day search impressions for declining pages: {avg_imp_declining:,.0f}")
```

## 4. Careful words: what I can and can't claim

**What the data supports:**
*   Observing historical patterns of traffic and engagement across different content types and age groups.
*   Identifying directional signals that suggest a page may be losing its search relevance.
*   Providing decision-support rankings to help editors prioritize their workflow.

**What I cannot claim (yet or ever):**
*   **Causation:** I cannot prove that a lack of updates *caused* a decline, nor that a refresh will *guarantee* a recovery.
*   **Predicting Google:** This work does not claim to reverse-engineer Google's proprietary ranking algorithms.
