## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## Research Question

**Decision:** Which content pages should be prioritized for refresh?

**Action:** The SEO or content team can update the pages predicted to decline before they lose more organic traffic.

**Cost of a wrong recommendation:** If a healthy page is selected unnecessarily, time and effort are wasted. If a declining page is missed, the website may continue losing traffic and potential visitors.

Machine learning can help by combining multiple signals such as impressions, CTR, page age, and update history to prioritize pages more effectively than a simple hand-written rule.
```
